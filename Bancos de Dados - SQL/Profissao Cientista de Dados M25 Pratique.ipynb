```python
import sqlite3 #Importando a biblioteca necessária
import pandas as pd
```

```python
df_vendas = pd.read_csv("TB_VENDAS_TAREFA.csv", delimiter=';')
```

```python
conn = sqlite3.connect(':memory:')
df_vendas.to_sql('tb_vendas', conn, index=False, if_exists='replace')
```

```python
def run_query(query):
    return pd.read_sql_query(query, conn)
```

# 1) Nomes dos produtos distintos


```python
query = "SELECT DISTINCT(PRODUTO) FROM TB_VENDAS"

result_df = run_query(query)
print(result_df)
```

# 2) Contagem dos clientes distintos

Dica: Você pode usar o distinct dentro do count()

```python
query = "SELECT COUNT(DISTINCT(PRODUTO)) AS QTD_PRODUTOS FROM TB_VENDAS"


result_df = run_query(query)
print(str(result_df.values[0][0]) + " produtos distintos na base de vendas.")
```

# 3) Produtos distintos onde o valor_unid é maior ou igual a 50 reais.

```python
query = "SELECT DISTINCT(PRODUTO), VALOR_UNID FROM TB_VENDAS WHERE VALOR_UNID >= 50"


result_df = run_query(query)
print(result_df)
```

# 4) ID das compras DAS 5 compras com mais valor total gasto

```python
query = "SELECT ID_COMPRA, VALOR_UNID*UNIDADES FROM TB_VENDAS ORDER BY VALOR_UNID*UNIDADES DESC LIMIT 5"

result_df = run_query(query)
print(result_df)
```

# 5) Podutos e preço da unidade dos produtos, ordenado do maior para o menor

```python

query = "SELECT DISTINCT(PRODUTO), VALOR_UNID FROM TB_VENDAS ORDER BY VALOR_UNID DESC"


result_df = run_query(query)
print(result_df)
```

# 6) Os 3 clientes que mais relalizaram compras e a quantidade de compras realizadas

```python
query = "SELECT ID_CLIENTE, COUNT(UNIDADES) FROM TB_VENDAS GROUP BY ID_CLIENTE ORDER BY COUNT(UNIDADES) DESC LIMIT 3"
result_df = run_query(query)
print(result_df)
```
