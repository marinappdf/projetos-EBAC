Para nossa atividade inicial de SQL trabalharemos com uma base de vendas.
Abaixo temos o ambiente de preparo para trabalharmos com o SQL lite.

```python
import sqlite3 #Importando a biblioteca necessária
import pandas as pd
```

```python
df_vendas = pd.read_csv("TB_VENDAS_TAREFA.csv", delimiter=';')
```

```python
conn = sqlite3.connect(':memory:')
df_vendas.to_sql('tb_vendas', conn, index=False, if_exists='replace')
```

```python
def run_query(query):
    return pd.read_sql_query(query, conn)
```

# 1) Consulta que retorna todos os dados da tabela de vendas:


```python
query = "SELECT * FROM TB_VENDAS"
result_df = run_query(query)
print(result_df)
```

# 2) Consulta que retorna apenas 10 linhas da coluna produto.


```python
query = "SELECT PRODUTO FROM TB_VENDAS limit 10"
result_df = run_query(query)
print(result_df)
```

# 3) Consulta que retorna a média do valor por unidade e a média das unidades vendidas.

```python
query = "SELECT AVG(VALOR_UNID) AS MEDIA_VALOR_UNID, AVG(UNIDADES) AS MEDIA_UNIDADES FROM TB_VENDAS"
result_df = run_query(query)
print(result_df)
```

# 4) consulta que retornA a coluna ID_COMPRA, ID_CLIENTE e o valor total gasto.


```python
query = "SELECT ID_COMPRA, ID_CLIENTE, VALOR_UNID*UNIDADES AS TOTAL FROM TB_VENDAS"
result_df = run_query(query)
print(result_df)
total = result_df['TOTAL']
```

# 5) Média do valor total gasto e média por pedido.

```python
print( "Média gasta pode venda\n R$ " + str(total.mean()))

query = "SELECT AVG(VALOR_UNID*UNIDADES) AS MEDIA_TOTAL FROM TB_VENDAS"
result_df = run_query(query)
print("Média de gastos por venda\n "+str(result_df['MEDIA_TOTAL'].values[0]))

query = "SELECT PRODUTO, AVG(VALOR_UNID*UNIDADES) AS MEDIA_GASTO_PRODUTO FROM TB_VENDAS GROUP BY PRODUTO"
result_df = run_query(query)
print("Média por Produto\n"+str(result_df))
```

# 6) Diferenças entra a aplicação do SQL e do Python para ciência de dados.

A linguagem SQL foi criada, e otimizada, para o acesso, edição e gerenciamento de dados em bancos de dados. Sua simplicidade não é fraqueza, ela facilita a aprendizagem e aplicação de suas ferramentas por aqueles que precisam dela.
Já o Python, em relação ao SQL, tem uma curva de aprendizado maior. Isso ocorre por que o python não foi feito para o gerenciamento de bancos de dados. Essa linguagem é muito mais ampla e poderosa, podendo ser usada para criação de jogos, desenvolvimento wev, dev ops, desenvolvimento de software, etc., e ciência de dados. Por isso, para aprender a usar ela para o gerenciamento de dados, é necessário aprender muitoos outros conceitos, como bibliotecas, lógica de programação, etc.

# 7) Diferenças entre o banco de dados relacionais de não relacionais.

Os bancos de dados relacionais são aqueles estruturados da forma mais convencional: tabelas com colunas e linhas. Seu aspecto relacional está na possibilidade de criar relações entre as tabelas por meio das chaves estrangeiras, das conexões entre campos de diferentes tabelas.
Os bancos não relacionais não estão organziados em colunas e linhas, estão organizados em pares chave-valor. Sua estrutura lembra a de objetos, ou listas e dicionários, em que um nome, chave, é dado para cada valor, ou conjunto de valores. Ao invés de colunas, os bancos tem chaves. Dentro dessas chaves há um valor que pode ser inteiro, texto, ou até mesmo objetos e arrays. Nos casos de serem objetos e arrays, eles são listas de valores, que podem também incluir novas listas de valores.

Enquanto nos bancos de dados relacionais é preciso que o número de linhas seja igual para todas as colunas, e vice-versa, o mesmo não é exigido dos bancos de dados relacionas: cada chave pode armazenar uma infinidade de valores independente do número de dados que as demais chaves possuem. 

# 8) O SQL pode ser mais prático que o Python para consultas rápidas?

Acho que essa pergunta não tem uma resposta absoluta. Tudo depende do contexto e situação em que isso se faz necessário. No caso de já ter sido construído ambientes de python próprios para o acesso e manipulação dos dados, aí realmente podemos dar preferência à ele. Caso contrário, o SQL é a  melhor escolha para consultas rápidas, pois facilita o manuseio de grandes bancos de dados.
